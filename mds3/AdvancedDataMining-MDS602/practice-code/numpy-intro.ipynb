```python
# Find and replace
import numpy as np
a = np.arange(-5, 5)
print(a)
print(np.where(a % 2 == 0)) # return index
print(a[np.where(a % 2 == 0)]) # return elements
print(np.where(a % 2 == 0, True, a)) # where with replace
```

```python
import matplotlib.pyplot as plt
def matrix_plot(m):
    fig, ax = plt.subplots()
    ax.imshow(m, cmap='Pastel1_r')
    shape = m.shape
    for i in range(m.size):
        y, x = divmod(i, shape[0])
        ax.text(x, y, i, size=12, ha='center', va='center')
    return fig
```

```python
# problem 1: in a 5*5 matrix from 0 to 24, highlight the numbers disivisble by 3
matrix = np.arange(25).reshape(5, 5)
div_by_3 = matrix % 3 == 0

matrix_plot(div_by_3)
```

```python
div_by_3_and_4 = (matrix % 3 == 0) & (matrix % 4 == 0)
print(div_by_3_and_4)
matrix_plot(div_by_3_and_4)
```

```python
img = plt.imread('../data/parrot.jpeg')
fig, ax = plt.subplots(layout='tight')
ax.imshow(img)
plt.show()

#  mask the image to display only the content in ellipse with center (300, 300) and axes (200, 100)
y, x, _ = img.shape
y_grid, x_grid = np.ogrid[:y, :x]
center = (300, 300)
axes = (200, 100)
mask = ((x_grid - center[0]) / axes[0]) ** 2 + ((y_grid - center[1]) / axes[1]) ** 2 <= 1
masked_img = np.zeros_like(img)
masked_img[mask] = img[mask]
fig, ax = plt.subplots(layout='tight')
ax.imshow(masked_img)
plt.show()
```
