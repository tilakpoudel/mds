```python
# Generate a 5 * 5 matrix from the numbers 1 to 25, by reshaping a list of numbers from 1 to 25. Highlight the grid to form a chess board like pattern, where the even numbers are highlighted.
import numpy as np
import matplotlib.pyplot as plt

# Create a list of numbers from 0 to 24
numbers = list(range(0, 25))
# Reshape the list into a 5x5 matrix
matrix = np.array(numbers).reshape(5, 5)
# Create a chess board pattern by highlighting odd numbers
chess_board = np.where(matrix % 2 == 1, 'X', 'O')

# Print the chess board pattern
# for row in chess_board:
#     print(' '.join(row))

# plot the matrix using matplotlib with red color for cells with 'X' and white color for cells with 'O', and add the numbers in each cell
plt.figure(figsize=(5, 5))


plt.imshow(chess_board == 'X', cmap='gray', vmin=0, vmax=1)
for i in range(5):
    for j in range(5):
        plt.text(j, i, str(matrix[i, j]), ha='center', va='center', color='red')
plt.title('Chess Board Pattern with Numbers')
plt.show()
```

```python
def matrix_plot(m):
    fig, ax = plt.subplots()
    ax.imshow(m, cmap='Pastel1_r')
    shape = m.shape
    for i in range(m.size):
        y, x = divmod(i, shape[0])
        ax.text(x, y, i, size=12, ha='center', va='center')
    return fig

m = np.zeros(25, dtype=np.int8).reshape(5, 5)
m[1, 1] = 1
fig = matrix_plot(m)
plt.show()
```

```python
m = np.zeros(25, dtype=np.int8).reshape(5, 5)

#  last row
m[-1, :] = 1
fig = matrix_plot(m)
plt.show()
```

```python
m = np.zeros(25, dtype=np.int8).reshape(5, 5)
#  second and fourth column
m[1:4, ::2] = 1

print(m)

fig = matrix_plot(m)
plt.show()

```

```python
m = np.zeros(25, dtype=np.int8).reshape(5, 5)
#  second and fourth column
m[...] = 0
m[1::2, 0::2] = 1

print(m)

fig = matrix_plot(m)
plt.show()
```
