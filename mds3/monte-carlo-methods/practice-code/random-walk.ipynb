```python
#  Generate a random walk in 1d for 100 steps and plot the result.

import numpy as np
import matplotlib.pyplot as plt

# Set the number of steps
num_steps = 10000
initial_state = 0
current_state = 0
positions = [initial_state]
displacements = []

for i in range(num_steps):
    #  Generate random numbers between 0 and 1
    random_number = np.random.rand()

    if (random_number == 0):
        pass
    elif random_number < 0.5:
        current_state -= 1
    else:
        current_state += 1
    
    displacements.append(current_state - initial_state)
    positions.append(current_state)

print('After {} steps, the final position is {}'.format(num_steps, current_state))
print('The positions at each step are: {}'.format(positions))

# Find the displacement from the initial position and calculate RMS displacement
final_displacement = current_state - initial_state

rms_displacement = np.sqrt(np.mean(np.array(displacements)**2))

print('Displacement from initial position: {}'.format(final_displacement))
print('RMS Displacement: {}'.format(rms_displacement))

# Plot the random walk
plt.figure(figsize=(10, 6))
plt.plot(positions)
plt.title('1D Random Walk')
plt.xlabel('Step')
plt.ylabel('Position')
plt.grid()
plt.show()
```

```python
def rms_displacement(displacements):
    return np.sqrt(np.mean(np.array(displacements)**2))
```

```python
# Create a random_walk function that takes the number of steps as an argument and returns the final position and 
# the list of positions and displacements at each step.
def random_walk(num_steps):
    initial_state = 0
    current_state = 0
    positions = [initial_state]
    displacements = []

    for i in range(num_steps):
        #  Generate random numbers between 0 and 1
        random_number = np.random.rand()

        if (random_number == 0):
            pass
        elif random_number < 0.5:
            current_state -= 1
        else:
            current_state += 1

        displacements.append(current_state - initial_state)
        positions.append(current_state)

    return {
        'current_state': current_state,
        'positions': positions,
        'displacements': displacements,
        'rms_displacement': rms_displacement(displacements)
    }
```

```python
# plot rms_displacement vs sqrt of number of steps
steps = [10000, 20000, 30000, 40000, 50000, 100000]

rms_displacements = []
plt.figure(figsize=(10,6))
for step in steps:
    result = random_walk(step)
    rms_displacements.append(result['rms_displacement'])
    # plot random walk for each step
    plt.plot(result['positions'])
    plt.title('1D Random Walk on various step')
    plt.xlabel('Step')
    plt.ylabel('Position')
    plt.grid()

plt.show()

# plot rms_displacements vs sqrt of steps
plt.figure(figsize=(10, 6))
plt.plot(np.sqrt(steps), rms_displacements, marker='o')
plt.title('RMS Displacement vs sqrt of Steps')
plt.xlabel('sqrt of Steps')
plt.ylabel('RMS Displacement')
plt.grid()
plt.show()
```
