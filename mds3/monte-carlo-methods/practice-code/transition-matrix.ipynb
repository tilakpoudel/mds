```python
#  Find the value of Pie on equilibrium state distribution in markov chain with transition matrix P
# Method 1: Using eigenvector corresponding to eigenvalue 1 start with equal probability and keep multiplying with P until convergence
import numpy as np

P = np.array([
    [0, 1],
    [1, 0]
])

n = len(P)

pi = np.ones(n) / n   # start with equal probability

for _ in range(1000):
    pi = np.dot(pi, P)

print("Stationary distribution:", pi)
```

```python
# solve for stationary distribution using eigenvector
# Method 2: Using eigenvector corresponding to eigenvalue 1
import numpy as np
P = np.array([
    [0, 1],
    [1, 0]
])
eigenvalues, eigenvectors = np.linalg.eig(P.T)
stationary_distribution = eigenvectors[:, np.isclose(eigenvalues, 1)]
stationary_distribution = stationary_distribution / np.sum(stationary_distribution)  # Normalize
print("Stationary distribution:", stationary_distribution.flatten())
```

```python
# General function to find stationary distribution for any transition matrix

import numpy as np

def find_stationary_distribution(P):
    """
    Find the stationary distribution (pi) of a Markov chain.

    Parameters
    ----------
    P : list or numpy array
        Transition matrix of the Markov chain (n x n)

    Returns
    -------
    pi : numpy array
        Stationary distribution vector
    """

    # Convert input to numpy array
    P = np.array(P)

    # Step 1: Transpose the matrix
    # Needed because eigenvector equation is: P^T * pi = pi
    PT = P.T

    # Step 2: Compute eigenvalues and eigenvectors
    eigenvalues, eigenvectors = np.linalg.eig(PT)

    # Step 3: Find index of eigenvalue closest to 1
    index = np.argmin(np.abs(eigenvalues - 1))

    # Step 4: Get the corresponding eigenvector
    stationary_vector = eigenvectors[:, index]

    # Step 5: Normalize so probabilities sum to 1
    pi = stationary_vector / stationary_vector.sum()

    # Remove tiny imaginary parts (numerical artifacts)
    return pi.real
```

```python
# P = [
#     [0.5,0.3,0.2],
#     [0.2,0.6,0.2],
#     [0.1,0.3,0.6]
# ]

P = np.array([
    [0, 1],
    [1, 0]
])

pi = find_stationary_distribution(P)

print(pi)
```
