```python
import numpy as np
```

```python

def class_integral(num=1000):
    x = np.random.rand(num)
    int_sum = 0 
    for i in x:
        int_sum += np.exp(-i**2)
    integral = (int_sum) / num
    return integral
```

```python
class_integral(10000)
```

```python
# Q2. Given function is: f(x) = exp(-2|x-5|). We need to plot this function for x in the range [0, 10].
# from numpy import sqrt

x = np.random.rand(100) * 10
fx = np.exp(-2 * abs(x-5))

py = 1 / (np.sqrt(2 * np.pi)) * np.exp(-0.5 * (x-5)**2)

fig, axs = plt.subplots(1, 2, figsize=(15, 5))
axs[0].scatter(x, fx)
axs[0].set_title('x vs fx')
axs[1].scatter(fx, py)
axs[1].set_title('y vs py')
```
