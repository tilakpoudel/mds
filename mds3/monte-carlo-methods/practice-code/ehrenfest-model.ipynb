```python
import random
import matplotlib.pyplot as plt

# Parameters
N = 100 # total number of balls
steps = 10000 # number of time steps

# Initial state: all balls in box A
k = N
history = []
balls_A = [k] # track number of balls in box A over time
balls_B = [0] # track number of balls in box B over time                                                                

for _ in range(steps):
    history.append(k)
    
    # Pick one ball at random
    if random.random() < k / N:
        k -= 1 # ball moved from A to B
        balls_A.append(k)
    else:
        k += 1 # ball moved from B to A
        balls_B.append(k)

# Plot the time evolution
plt.figure(figsize=(10, 4))
plt.plot(history, alpha=0.7)
# plt.axhline(N/2, color='red', linestyle='--', label='Equilibrium (N/2)')
plt.xlabel('Time step')
plt.ylabel('Balls in Box A')
plt.title('Ehrenfest Model Simulation')
plt.legend()
plt.show()

# Plot the distribution of balls in box A over time
plt.figure(figsize=(10, 4))
plt.hist(history, bins=range(N+2), density=True, alpha=0.7)
plt.axvline(N/2, color='red', linestyle='--', label='Equilibrium (N/2)')
plt.xlabel('Number of Balls in Box A')
plt.ylabel('Probability')
plt.title('Distribution of Balls in Box A')
plt.legend()
plt.show()

#  plot number of balls in box A and box B over time
plt.figure(figsize=(10, 4))
plt.plot(balls_A, label='Box A', alpha=0.7)
plt.plot(balls_B, label='Box B', alpha=0.7)
plt.axhline(N/2, color='red', linestyle='--', label='Equilibrium (N/2)')
plt.xlabel('Time step')
plt.ylabel('Number of Balls')
plt.title('Ehrenfest Model: Balls in Box A and B Over Time')
plt.legend()
plt.show()
```

```python
# Alternative algorithm for Ehrenfest model simulation:
# Number ball from 1 to 100
# Randomly choose a number from 1 to 100
# If the number is in box A, move it to box B; if it's in box B, move it to box A

def simulate_ehrenfest(N, steps):
    # Initialize boxes
    box_A = set(range(1, N+1)) # all balls start in box A
    box_B = set() # box B starts empty
    
    history_A = []
    history_B = []
    
    for _ in range(steps):
        # Randomly choose a ball
        ball = random.randint(1, N)
        
        if ball in box_A:
            box_A.remove(ball)
            box_B.add(ball)
        else:
            box_B.remove(ball)
            box_A.add(ball)
        
        history_A.append(len(box_A))
        history_B.append(len(box_B))
    
    return history_A, history_B

# Run the simulation
N = 100
steps = 10000
history_A, history_B = simulate_ehrenfest(N, steps)

# Plot the results
plt.figure(figsize=(10, 4))
plt.plot(history_A, label='Box A', alpha=0.7)
plt.plot(history_B, label='Box B', alpha=0.7)
plt.axhline(N/2, color='red', linestyle='--', label='Equilibrium (N/2)')
plt.xlabel('Time step')
plt.ylabel('Number of Balls')
plt.title('Ehrenfest Model Simulation (Alternative Algorithm)')
plt.legend()
plt.show()

# plot number of balls in box A over N
plt.figure(figsize=(10, 4))
plt.hist(history_A, bins=range(N+2), density=True, alpha=0.7)
plt.axvline(N/2, color='red', linestyle='--', label='Equilibrium (N/2)')
plt.xlabel('Number of Balls in Box A')
plt.ylabel('Probability')
plt.title('Distribution of Balls in Box A (Alternative Algorithm)')
plt.legend()
plt.show()
```
