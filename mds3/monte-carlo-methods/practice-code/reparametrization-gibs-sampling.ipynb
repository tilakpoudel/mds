```python
# We want to estimate the joint distribution of x and y where x : uniform(-2,2) and y = x^2
import numpy as np

# Gibbs sampling for a bivariate normal distribution
def gibbs_sampling(row = 0.5, num_samples = 1000, x = 0, y = 0):
    samples = np.zeros((num_samples, 2))
    x, y = x, y  # Initial values

    for i in range(num_samples):
        x = np.random.uniform(-2, 2)  # Sample x from uniform distribution
        y = x**2  # Sample y from the deterministic relationship y = x^2
        samples[i] = [x, y]

    return samples
```

```python
# Generate samples
number_of_samples = 1000
# x = np.random.uniform(-2, 2, number_of_samples)
# y = x**2

samples_without_reparameterization = gibbs_sampling(num_samples=number_of_samples)

print("Samples without reparameterization:")
# print(samples_without_reparameterization)

# plotting the samples
import matplotlib.pyplot as plt

# plot the histogram of the samples
plt.hist(samples_without_reparameterization[:, 1], bins=30)
plt.title('Histogram of Samples without Reparameterization')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()

# scatter plot of the samples
plt.scatter(samples_without_reparameterization[:, 0], samples_without_reparameterization[:, 1], alpha=0.5)
plt.title('Scatter Plot of Samples without Reparameterization')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```

```python
#  with reparameterization, initialize starting values for u and v using uniform distribution in the range [0,1]
# reparameterize u and v to the range of x and y respectively, x from (4*u-2 and y from (2*v-1)*x)
# Implement gibbs sampling iterations:
# 1. Sample u from its conditional distribution given v, which is uniform(0,1)
# 2. Sample v from its conditional distribution given u, which is uniform(0,1)
# 3. Reparameterize u and v to to the range of x and y respectively, x from (4*u-2 and y from (2*v-1)*x)
# 4. Store the samples in the respective arrays

# Gibbs sampling with reparameterization, Not sure if actual gibbs sampling is happening here, but we are generating samples using the reparameterization technique
def gibbs_sampling_reparameterized(num_samples = 1000, u = 0.5, v = 0.5):
    samples = np.zeros((num_samples, 2))
    u, v = u, v  # Initial values

    for i in range(num_samples):
        u = np.random.uniform(0, 1)  # Sample u from uniform distribution
        v = np.random.uniform(0, 1)  # Sample v from uniform distribution
        x = 4 * u - 2  # Reparameterize u to the range of x
        y = (2 * v - 1) * x  # Reparameterize v to the range of y
        samples[i] = [x, y]

    return samples
```

```python
# Plot in histogram the samples obtained from the reparameterized Gibbs sampling
number_of_samples = 10000
samples_with_reparameterization = gibbs_sampling_reparameterized(num_samples=number_of_samples)
# print("Samples with reparameterization:")
# print(samples_with_reparameterization)

plt.hist(samples_with_reparameterization[:, 1], bins=30)
plt.title('Histogram of Samples with Reparameterization')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()

# plot the scatter plot of the samples obtained from the reparameterized Gibbs sampling
plt.scatter(samples_with_reparameterization[:, 0], samples_with_reparameterization[:, 1], alpha=0.5)
plt.title('Scatter Plot of Samples with Reparameterization')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```
