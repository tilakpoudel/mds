```python
#  Acceptance and rejection sampling

import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 1.2- x**4

n = 1000
xs = np.linspace(0, 1, n)
ys = f(xs)

#  plot the function
plt.plot(xs, ys)
plt.title('f(x) vs x')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.show()
```

```python
# acceptance-rejection sampling
accepted_points_x = []
rejection_points_x = []

accepted_points_y = []
rejection_points_y = []
total_points = 0

n = 1000

for i in range(n):
    x = np.random.uniform(0, 1)
    y = np.random.uniform(0, 1.2)  # max value of f(x) is 1.2

    if y < f(x):
        accepted_points_x.append(x)
        accepted_points_y.append(y)
    else:
        rejection_points_x.append(x)
        rejection_points_y.append(y)

#  plot the accepted and rejected points
plt.scatter(accepted_points_x, accepted_points_y, color='g', label='Accepted Points')
plt.scatter(rejection_points_x, rejection_points_y, color='r', label='Rejected Points')
plt.title('Accepted and Rejected Points')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()

accepted_points_count = len(accepted_points_x)
print(f'Number of accepted points: {accepted_points_count}')

#  Find the value of intergral of f(x) from 0 to 1
integral_estimate = accepted_points_count / n * 1.2  # area
acceptance_rate = accepted_points_count / n
print(f'Estimated integral of f(x): {integral_estimate}')
print(f'Acceptance rate: {acceptance_rate}')
```
