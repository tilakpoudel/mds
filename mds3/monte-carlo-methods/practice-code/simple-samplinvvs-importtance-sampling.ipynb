```python
# Integrate e**x from 0 to 1 using Monte Carlo method
import numpy as np
import matplotlib.pyplot as plt
import random

# Generate random numbers using built-in RNG
def built_in_rng(seed, n):
    """Generate n pseudo-random numbers using built-in RNG."""
    np.random.seed(seed)

    return np.random.rand(n)

# Given function to integrate
def f(x):
    return np.exp(-x ** 2)

def monte_carlo_integration(fx, a=0, b=1):
    """Estimate the integral using Monte Carlo method."""
    mean=  np.mean(fx)

    i = mean / (b - a)

    return i

#  Calculate the error of the integral estimate
def absolute_error(estimate, true_value):
    return abs(estimate - true_value)

#  Calculate the standard error of the estimate
def standard_error(fx):
    return np.std(fx) / np.sqrt(len(fx))

# Generate random numbers using LCG and built-in RNG
N = 1000
seed = 58942115
x2 = built_in_rng(seed=seed, n=N)
#  Calculate the function values for both sets of random numbers
f_x2 = f(x2)

# Integration is given by the mean value of f(x) over [0, 1]
integral_x2 = monte_carlo_integration(f_x2)

print(f"Integral estimate using built-in RNG: {integral_x2:.8f}")

# print(f"x1 (LCG): Mean = {np.mean(x1):.8f}, Std Dev = {np.std(x1):.8f}")
# print(f" Mean of f(x2): {np.mean(f_x2):.4f}")

# Display the error and standard error for both methods
true_value = np.exp(-1) - 1  # True value of the integral
print(f"True value of the integral: {true_value:.8f}")
error_x2 = absolute_error(integral_x2, true_value)
std_error_x2 = standard_error(f_x2)

print(f"Error of integral estimate using built-in RNG: {error_x2:.8f}")
print(f"Standard error of estimate using built-in RNG: {std_error_x2:.8f}")


print("*" * 20 )
print("Importance sampling")
# Using importance sampling
# Function f(y)
def f_y(x):
    return (np.exp(1) / (np.exp(1) - 1)) * (1 - np.exp(-x))

# Function to integrate
def fun_to_integrate(x):
    return (np.exp(-f_y(x)**2)) / (np.exp(-f_y(x)))

def importance_sampling():
    i = 0
    sum = 0
    summ= 0

    while i <=N:
        x = random.random
        y = ((np.exp(1) / (np.exp(1) - 1)) * (1 - np.exp(-x)))
        f = fun_to_integrate(y)
        sum = sum + f
        summ = summ + f*f
        i= i +1
    
    meanf = sum / N
    MCI = meanf * (np.exp(1) - np.exp(-1))
    std_MCI = summ / (N - meanf**2)
    print('Integral value using importance sampling with MC' + str(MCI))
    print('Standard deviation of monte carlo integration' + str(std_MCI))

# N = 1000
importance_sampling()
# seed = 589421
# x3 = built_in_rng(seed=seed, n=N)
# #  Calculate the function values for both sets of random numbers
# f_x3 = fun_to_integrate(x3)

# # Integration is given by the mean value of f(x) over [0, 1] using the importance sampling
# integral_x3_importance_sampling = monte_carlo_integration(f_x3)

# print(f"Integral estimate using importance sampling: {integral_x3_importance_sampling:.8f}")
```
